# Data Preparation

This notebook checks the data before model training.

We will:

- load the IoT clean data
- display the IoT clean data
- check IoT null values
- load the Blattmann clean data
- display the Blattmann clean data
- check Blattmann null values
- check the columns and class labels we will use later


#### 1. Imports


````python
# Import all libraries in one place.
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from pathlib import Path

pd.set_option('display.max_columns', None)
pd.set_option('display.width', 120)
sns.set_theme(style='whitegrid')
````

#### 2. File Paths


````python
# Define the files we will use.
data_folder = Path('.')

iot_file = data_folder / 'iot_clean_data.xlsx'
blattmann_file = data_folder / 'blattmann_clean_data.xlsx'

features = ['temperature', 'humidity', 'tvoc_ppb', 'eco2_ppm']
target_column = 'class_id'
class_names = ['Normal', 'Cooking', 'Fire']
````

#### 3. Load and Display IoT Clean Data

The IoT data contains the real device readings. This data is important because it includes the `Cooking` class.


````python
# Load and display the IoT clean data.
iot_data = pd.read_excel(iot_file)

print('IoT data shape:', iot_data.shape)
print('IoT columns:')
print(iot_data.columns.tolist())

display(iot_data.head())
````

#### 4. Check Null Values in IoT Data

Null values are missing values. We check them because missing values can break model training or make the model learn wrong patterns.


````python
# Check null values in the IoT data.
iot_null_values = iot_data.isnull().sum()

display(iot_null_values.to_frame('null_count'))

print('Total IoT null values:', iot_null_values.sum())
````

#### 5. Load and Display Blattmann Clean Data

The Blattmann/mock data contains the larger smoke dataset. In our project mapping, it gives us `Normal` and `Fire` rows.


````python
# Load and display the Blattmann clean data.
blattmann_data = pd.read_excel(blattmann_file, sheet_name='Blattmann Processed')

print('Blattmann data shape:', blattmann_data.shape)
print('Blattmann columns:')
print(blattmann_data.columns.tolist())

display(blattmann_data.head())
````

#### 6. Check Null Values in Blattmann Data

We check missing values in the Blattmann data before using it for model training.


````python
# Check null values in the Blattmann data.
blattmann_null_values = blattmann_data.isnull().sum()

display(blattmann_null_values.to_frame('null_count'))

print('Total Blattmann null values:', blattmann_null_values.sum())
````

#### 7. Check Columns We Need

For the model, we want to use the same feature names in both datasets.


````python
# Check if both datasets contain the columns we need.
needed_columns = features + ['label', target_column]

print('Needed columns:')
print(needed_columns)

print('\nMissing columns in IoT data:')
print([column for column in needed_columns if column not in iot_data.columns])

print('\nMissing columns in Blattmann data:')
print([column for column in needed_columns if column not in blattmann_data.columns])
````

#### 8. Check Class Distribution

This shows how many rows we have for each class.

Class mapping:

```text
0 = Normal
1 = Cooking
2 = Fire
```


````python
# Check how many rows exist for each class in each dataset.
print('IoT class distribution:')
display(iot_data['label'].value_counts().to_frame('count'))

print('\nBlattmann class distribution:')
display(blattmann_data['label'].value_counts().to_frame('count'))
````
