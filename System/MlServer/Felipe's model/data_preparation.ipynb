```python
import warnings; warnings.filterwarnings('ignore')
from pathlib import Path
import pandas as pd

NOTEBOOK_DIR = Path('.').resolve()
FEATURES = ['temperature', 'humidity', 'tvoc_ppb', 'eco2_ppm']
```

```python
iot = pd.read_excel(NOTEBOOK_DIR / 'iot_labeled.xlsx')
print(f"IoT shape: {iot.shape}")
print(iot['label'].value_counts().to_string())
print(iot[FEATURES].describe().round(2))
```

```python
blatt = pd.read_excel(NOTEBOOK_DIR / 'blattmann_processed.xlsx', sheet_name='Blattmann Processed')
print(f"Blattmann shape: {blatt.shape}")
print(blatt['label'].value_counts().to_string())
print(blatt[FEATURES].describe().round(2))
```

```python
# missing values
for name, df in [('IoT', iot), ('Blattmann', blatt)]:
    nulls = df[FEATURES].isnull().sum()
    print(f"{name} nulls:", nulls[nulls > 0].to_dict() if nulls.any() else "none")
```

```python
# outlier check — ENS160 max range is 65,000 ppb/ppm
for name, df in [('IoT', iot), ('Blattmann', blatt)]:
    t = (df['tvoc_ppb'] > 65_000).sum()
    e = (df['eco2_ppm'] > 65_000).sum()
    print(f"{name} — tvoc outliers: {t}  eco2 outliers: {e}")
```
